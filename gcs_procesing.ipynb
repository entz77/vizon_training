```python
from google.cloud import storage

PROJECT_ID = "vveesee"
gcs_uri = "gs://prd-vveesee-edge-submit/images/"

bucket_name, _, prefix = gcs_uri[len("gs://"):].partition("/")
prefix = prefix.strip("/")

client = storage.Client(project=PROJECT_ID)
blobs = client.list_blobs(bucket_name, prefix=f"{prefix}/" if prefix else None)

object_names = [blob.name for blob in blobs]
print(f"Found {len(object_names)} objects under gs://{bucket_name}/{prefix}")
object_names
```

```python
sel_objects = []
for obj in object_names:
    if obj.endswith(".jpg") and f"uta00" in obj and '_capture_' not in obj and '_202605' not in obj:
        ob = obj.split("/")[-1]
        sel_objects.append(ob)

print(len(sel_objects))
sel_objects[0]
```

```python
import pandas as pd
excl = pd.read_csv("malang_objects_labeled.csv")
excl = excl['img'].to_list()
print(len(excl))
excl[0]
```

```python
sel_objects = [obj for obj in sel_objects if obj not in excl]
len(sel_objects)
```

```python
final_sel_objects = []
for obj in sel_objects:
    obj = f'gs://prd-vveesee-edge-submit/images/{obj}'
    final_sel_objects.append(obj)

with open('malang_tobe_infered.txt', 'w') as f:
    for item in final_sel_objects:
        f.write("%s\n" % item)
```
